# Nitrogen × Spacing Interaction (Factorial Experiment)

This experiment studies whether crop response to nitrogen fertilizer depends on plant spacing.

Objective:
Determine if management factors interact and identify the best treatment combination.

```python
#Import the packagess
import pandas as pd
from agrodesign.experiment import Experiment
```

```python
#Dataset
df = pd.DataFrame({
    "Rep":[1]*6 + [2]*6 + [3]*6,
    "Nitrogen":["Low","Low","Medium","Medium","High","High"]*3,
    "Spacing":["Narrow","Wide"]*9,
    "Yield":[
        42,46,50,55,58,64,
        43,45,49,54,57,63,
        41,47,51,56,59,65
    ]
})

df
```

```python
#Run Analysis
result = Experiment(df,"Yield").factorial(["Nitrogen","Spacing"]).run()
result
```

```python
#Full Report
print(result)
```

```python
#Biological interpretation
result.summary()

#👉 This is the key learning moment:
#AgroDesign should recommend a combination, not a single factor.
```

```python
#Visual Inspection
result.plot()
```

```python
#Export
result.export("interaction_report")
```
