# Fertilizer Evaluation Using RCBD

This experiment evaluates four fertilizer treatments in a field trial using a Randomized Complete Block Design (RCBD).

Objective:
Determine which fertilizer produces the highest grain yield while accounting for field variability.


```python
#Import the packages
import pandas as pd
from agrodesign.experiment import Experiment
```

```python
#Dataset
df = pd.DataFrame({
    "Block":["B1","B1","B1","B1",
             "B2","B2","B2","B2",
             "B3","B3","B3","B3",
             "B4","B4","B4","B4"],
    "Fertilizer":["Control","NPK_50","NPK_100","Organic"]*4,
    "Yield":[41,52,60,48,43,55,62,47,40,53,59,50,42,54,61,49]
})

df
```

```python
#Run Analysis
result = Experiment(df,"Yield").rcbd("Fertilizer","Block").run()
result
```

```python
#Get Full Statistical Report
print(result)
```

```python
#Agronomic Interpretation
result.summary()
```

```python
#Visualization
result.plot()
```

```python
#Export the Files
result.export("fertilizer_report")
```
